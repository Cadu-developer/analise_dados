```python
import pandas as pd
```

```python

path = input("Digite o caminho do arquivo Excel: ")
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar

```python

formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
formacao_total.columns = ["uf_regiao", "total", "total_homens", "total_mulheres"]

formacao_total["percentual_mulheres"] = ((formacao_total["total_mulheres"] / formacao_total["total"]) * 100).round(2)
formacao_ctem = formacao_bruto.iloc[3:36, 4:8].copy()
formacao_ctem.columns = ["uf_regiao", "total", "total_homens", "total_mulheres"]
formacao_ctem["percentual_mulheres"] = ((formacao_ctem["total_mulheres"] / formacao_ctem["total"]) * 100).round(2)

formacao_educacao_saude = formacao_bruto.iloc[3:36, 8:12].copy()
formacao_educacao_saude.columns = ["uf_regiao", "total", "total_homens", "total_mulheres"]
formacao_educacao_saude["percentual_mulheres"] = ((formacao_educacao_saude["total_mulheres"] / formacao_educacao_saude["total"]) * 100).round(2)

print("=== Curso superior ===")
display(formacao_total)

print("=== Ciência, Tecnologia, Engenharias e Matemática ===")
display(formacao_ctem)

print("=== Educação, Serviços pessoais, Saúde e Bem-estar ===")
display(formacao_educacao_saude)
```

### Resultado
A tabela original foi dividida em três tabelas conforme solicitado: curso superior; Ciência, Tecnologia, Engenharias e Matemática; e Educação, Serviços pessoais, Saúde e Bem-estar.
